```python
import pandas as pd
import numpy as np
```

Pandas Series

```python
#creating a series from a list
nums=['apple','banana','cherry','date','fig','grape']
nums_series=pd.Series(nums)
print(nums_series)
```

```python
#adding index to the series
index_series=['first','second','third','fourth','fifth','sixth']
nums_series=pd.Series(nums,index=index_series)
print(nums_series)
```

```python
#setting a name to the series
nums_series.name='Fruits'
print(nums_series)
```

```python
#creating a series from a dictionary
nums_dict={'first':'apple','second':'banana','third':'cherry','fourth':'date','fifth':'fig','sixth':'grape'}
nums_series=pd.Series(nums_dict)
print(nums_series)
```

```python
#getting the size of the series
print(nums_series.size)
```

```python
#find the index of a specific value in the series
index_of_value=nums_series[nums_series=='cherry'].index[0]
print(index_of_value)
```

```python
#slicing the series
print(nums_series[1:4])
```

```python
#performing arthemetic operations on series
nums_series1=pd.Series([1,2,37,4,5])
nums_series2=pd.Series([10,20,30,40,50])
print(nums_series1+nums_series2)
```

```python
#relational operations on series
print(nums_series1>nums_series2)
```
