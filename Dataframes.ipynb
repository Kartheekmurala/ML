DataFrame

```python
import pandas as pd
import numpy as np
```

```python
#using a list of lists to create a dataframe
student_data = [
   [1001,20,85],
   [1002,22,90],
   [1003,23,78],
   [1004,21,95]
]
df = pd.DataFrame(student_data,columns=['ID', 'Age', 'Grade'])
print(df)
```

```python
#using a dictionary to create a dataframe
student_data = {
    "ID":[1001,1002,1003,1004],
    "Age":[20,22,23,21],
    "Grade":[85,90,78,95]
    }
df = pd.DataFrame(student_data)
print(df)
```

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

df = kagglehub.load_dataset(
    KaggleDatasetAdapter.PANDAS,
    "prakashmahatra/top-rated-movies",
    "movies.csv"      # actual file name
)
#printing first 5 rows of the dataframe
print(df.head())
```

```python
#printing the shape of the dataframe
print(df.shape,end="\n\n")
#printing the column names of the dataframe
print(df.columns,end="\n\n")
#printing the summary of the dataframe
print(df.info())
```

```python
#Descriptive statistics of the dataframe
print(df.describe())
```

```python
#print head of the dataframe
print("PRINTING HEAD OF THE DATAFRAME")
print(df.head(),end="\n\n")
print("PRINTING TAIL OF THE DATAFRAME")
#print tail of the dataframe
print(df.tail())
```
